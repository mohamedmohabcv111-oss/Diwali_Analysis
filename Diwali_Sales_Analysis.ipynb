```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
```

```python
df = pd.read_csv('Diwali Sales Data.csv', encoding='unicode_escape')

df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
df.head()
```

```python
pd.isnull(df).sum()
```

```python
df = df.dropna(subset=['Amount'])
```

```python
pd.isnull(df).sum()
```

```python
df.head()
```

```python
df.info()
```

```python
df['Amount']  = df['Amount'].astype('int64')
```

```python
df.info()
```

```python
df[['Age','Orders','Amount']].describe()
```

```python
df.columns
```

```python
ax = sns.countplot(x = 'Gender' , data= df)

for bars in ax.containers:
    ax.bar_label(bars)
    
```

```python
sales_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sales_plot = sns.barplot(x='Gender', y='Amount', data=sales_gen)

for bars in sales_plot.containers:
    sales_plot.bar_label(bars, fmt='{:,.0f}', padding=3)

sales_plot.yaxis.set_major_formatter('{x:,.0f}')
plt.ylim(0, sales_gen['Amount'].max() * 1.15)
```

From the above graph we can conclude that most of the purchasing power is from females

```python
ax = sns.countplot(data = df, x = 'Age Group', hue = 'Gender')

for bar in ax.containers:
    ax.bar_label(bar)
```

```python
sales_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sales_age_plot = sns.barplot(x='Age Group', y='Amount', data=sales_age)

for bars in sales_age_plot.containers:
    sales_age_plot.bar_label(bars, fmt='{:,.0f}', padding=3)

sales_age_plot.yaxis.set_major_formatter('{x:,.0f}')
plt.ylim(0, sales_age['Amount'].max() * 1.15)
```

From above graphs we can see most of the buyers are females of age group 26-35yrs

```python
states_orders = df.groupby('State').size().sort_values(ascending=False).to_frame('Orders')

sns.heatmap(states_orders,annot=True,fmt='d',cmap='YlOrRd');
plt.title('Number Of Orders by State')
plt.show()
```

Plot above shows the amount of orders for each state

```python
sns.set_theme(style='whitegrid')

top10 = df.groupby('State')['Amount'].sum().sort_values(ascending=False).head(10)

fig, ax = plt.subplots(figsize=(12, 6))

top10.plot(kind='bar', color='steelblue', edgecolor='black', ax=ax)

ax.set_title('Top 10 States by Total Amount Spent', fontsize=16, fontweight='bold')
ax.set_xlabel('State', fontsize=12)
ax.set_ylabel('Total Amount', fontsize=12)


ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x:,.0f}'))


for container in ax.containers:
    ax.bar_label(container, labels=[f'{v:,.0f}' for v in top10], fontsize=9, padding=3)

plt.xticks(rotation=45, ha='right')
sns.despine()
plt.tight_layout()
plt.show()
```

Plot above thows the total amount the top 10 states spend - most orders are from(Uttar Pradesh , Maharashtra & Karnataka)

```python
counts = df['Zone'].value_counts()

plt.pie(counts, labels=counts.index, autopct='%1.1f%%', wedgeprops={'width': 0.6})
plt.title('Share of Orders by Zone')
plt.show()

```

The plot shows each zone's percentage of total orders. The West zone accounts for the largest share , while the East zone has the smallest .

```python
table = df.pivot_table(index='State', columns='Product_Category', values='Amount', aggfunc='sum', fill_value=0)

top_states = table.sum(axis=1).sort_values(ascending=False).head(10).index
top_cats = table.sum(axis=0).sort_values(ascending=False).head(8).index  
table = table.loc[top_states, top_cats]

plt.figure(figsize=(12, 7))
sns.heatmap(table, annot=True, fmt='.0f', cmap='YlOrRd')
plt.title('Total Amount by State and Product Category')
plt.show()
```
